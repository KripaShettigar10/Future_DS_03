```python
import pandas as pd

df = pd.read_csv("e_commerce_oct_2019.csv",nrows=100000)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Check column names
print("Column Names:")
print(df.columns.tolist())

# Check dataset information
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check missing values again
print("\nMissing values after checking duplicates:")
print(df.isnull().sum())

# Check dataset shape
print("\nDataset shape after removing duplicates:", df.shape)

# Display the first five rows
df.head()
```

```python
# Count each type of customer event
print("Event Types:")
print(df['event_type'].value_counts())

# Calculate the percentage of each event type
print("\nEvent Percentages:")
print((df['event_type'].value_counts(normalize=True) * 100).round(2))

# Display the first five rows
df.head()
```

```python
# Display all column names
print("Dataset Columns:")
print(df.columns.tolist())
```

```python
# Count unique users at each funnel stage

views = df[df['event_type'] == 'view']['user_id'].nunique()

carts = df[df['event_type'] == 'cart']['user_id'].nunique()

purchases = df[df['event_type'] == 'purchase']['user_id'].nunique()

# Calculate conversion rates
view_to_cart = (carts / views) * 100 if views else 0

cart_to_purchase = (purchases / carts) * 100 if carts else 0

view_to_purchase = (purchases / views) * 100 if views else 0

# Display results
print("MARKETING FUNNEL ANALYSIS")
print("-------------------------")
print("Unique users who viewed products:", views)
print("Unique users who added to cart:", carts)
print("Unique users who purchased:", purchases)

print("\nCONVERSION RATES")
print("-------------------------")
print(f"View to Cart: {view_to_cart:.2f}%")
print(f"Cart to Purchase: {cart_to_purchase:.2f}%")
print(f"View to Purchase: {view_to_purchase:.2f}%")
```

```python
# Identify users who performed each event
view_users = set(df.loc[df['event_type'] == 'view', 'user_id'].dropna())
cart_users = set(df.loc[df['event_type'] == 'cart', 'user_id'].dropna())
purchase_users = set(df.loc[df['event_type'] == 'purchase', 'user_id'].dropna())

# Find users shared between funnel stages
view_and_cart = view_users & cart_users
cart_and_purchase = cart_users & purchase_users
view_and_purchase = view_users & purchase_users

# Users who performed all three events
all_three = view_users & cart_users & purchase_users

print("USER OVERLAP ANALYSIS")
print("-------------------------")
print("Users who viewed and added to cart:", len(view_and_cart))
print("Users who added to cart and purchased:", len(cart_and_purchase))
print("Users who viewed and purchased:", len(view_and_purchase))
print("Users who performed all three events:", len(all_three))
```

```python
# Calculate user overlap conversion rates
view_to_cart_overlap = (
    len(view_and_cart) / len(view_users) * 100
)

cart_to_purchase_overlap = (
    len(cart_and_purchase) / len(cart_users) * 100
)

view_to_purchase_overlap = (
    len(view_and_purchase) / len(view_users) * 100
)

print("USER OVERLAP CONVERSION RATES")
print("-----------------------------")
print(f"View and Cart / View Users: {view_to_cart_overlap:.2f}%")
print(f"Cart and Purchase / Cart Users: {cart_to_purchase_overlap:.2f}%")
print(f"View and Purchase / View Users: {view_to_purchase_overlap:.2f}%")
```

```python
# Check the most active product categories

category_analysis = df.groupby(
    ['category_code', 'event_type']
).size().unstack(fill_value=0)

# Display the top 10 categories by purchase events
if 'purchase' in category_analysis.columns:
    top_categories = category_analysis.sort_values(
        by='purchase', ascending=False
    ).head(10)

    print("TOP 10 CATEGORIES BY PURCHASE EVENTS")
    display(top_categories)
else:
    print("Purchase event data is unavailable.")
```

```python
# Calculate category-level event counts
category_events = df.pivot_table(
    index='category_code',
    columns='event_type',
    values='product_id',
    aggfunc='count',
    fill_value=0
)

# Keep categories with both views and purchases
category_events = category_events[
    (category_events['view'] > 0) &
    (category_events['purchase'] > 0)
].copy()

# Calculate purchase events relative to view events
category_events['purchase_view_ratio'] = (
    category_events['purchase'] /
    category_events['view'] * 100
)

# Display top 10 categories by purchase-to-view ratio
top_conversion_categories = category_events.sort_values(
    by='purchase_view_ratio',
    ascending=False
).head(10)

print("TOP 10 CATEGORIES BY PURCHASE-TO-VIEW EVENT RATIO")
display(
    top_conversion_categories[
        ['view', 'purchase', 'purchase_view_ratio']
    ].round(2)
)
```

```python
print("Current dataset shape:", df.shape)
print("Unique sessions:", df['user_session'].nunique())
print("\nEvent counts:")
print(df['event_type'].value_counts())
```

```python
# Count sessions containing each event type
session_summary = df.groupby('user_session')['event_type'].agg(
    lambda events: set(events)
)

total_sessions = len(session_summary)

view_sessions = session_summary.apply(lambda x: 'view' in x).sum()
cart_sessions = session_summary.apply(lambda x: 'cart' in x).sum()
purchase_sessions = session_summary.apply(lambda x: 'purchase' in x).sum()

print("Total Sessions:", total_sessions)
print("Sessions with Product Views:", view_sessions)
print("Sessions with Cart Additions:", cart_sessions)
print("Sessions with Purchases:", purchase_sessions)

print("\nSESSION-LEVEL EVENT RATES")
print("Cart / View:", round(cart_sessions / view_sessions * 100, 2), "%")
print("Purchase / View:", round(purchase_sessions / view_sessions * 100, 2), "%")
print("Purchase / Cart:", round(purchase_sessions / cart_sessions * 100, 2), "%")
```

```python
# Sort events by time
df['event_time'] = pd.to_datetime(df['event_time'])
df = df.sort_values('event_time')

# Identify sessions containing events
session_events = df.groupby('user_session').agg(
    events=('event_type', list)
)

# Count sessions that followed the funnel in order
view_count = 0
cart_after_view = 0
purchase_after_cart = 0

for events in session_events['events']:
    if 'view' in events:
        view_count += 1

        view_position = events.index('view')

        if 'cart' in events[view_position + 1:]:
            cart_after_view += 1

            cart_position = events.index('cart', view_position + 1)

            if 'purchase' in events[cart_position + 1:]:
                purchase_after_cart += 1

print("Sessions with Views:", view_count)
print("Sessions with View → Cart:", cart_after_view)
print("Sessions with View → Cart → Purchase:", purchase_after_cart)

if view_count > 0:
    print("View to Cart Rate:",
          round(cart_after_view / view_count * 100, 2), "%")

if cart_after_view > 0:
    print("Cart to Purchase Rate:",
          round(purchase_after_cart / cart_after_view * 100, 2), "%")
```

```python
import matplotlib.pyplot as plt

stages = ['Product Views', 'Added to Cart', 'Purchased']
counts = [24376, 798, 437]

plt.figure(figsize=(8, 5))
bars = plt.bar(stages, counts)

plt.title('E-commerce Marketing Funnel Analysis')
plt.xlabel('Funnel Stage')
plt.ylabel('Number of Sessions')

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f'{int(height):,}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.savefig('funnel_analysis.png', dpi=300)
plt.show()
```

```python
# Remove rows without category information
category_df = df.dropna(subset=['category_code']).copy()

# Count events for each category
category_analysis = category_df.pivot_table(
    index='category_code',
    columns='event_type',
    values='product_id',
    aggfunc='count',
    fill_value=0
)

# Find the top 10 categories by purchase events
top_categories = category_analysis.sort_values(
    by='purchase',
    ascending=False
).head(10)

print("Top 10 Product Categories by Purchase Events:")
display(top_categories)
```

```python
import matplotlib.pyplot as plt

purchase_data = top_categories.sort_values(
    by='purchase',
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    purchase_data.index,
    purchase_data['purchase']
)

plt.title('Top 10 Product Categories by Purchase Events')
plt.xlabel('Number of Purchase Events')
plt.ylabel('Product Category')

plt.tight_layout()
plt.savefig('top_product_categories.png', dpi=300)
plt.show()
```

```python
# Calculate purchase-to-view event ratio for each category
category_analysis['purchase_view_ratio'] = (
    category_analysis['purchase'] /
    category_analysis['view'].replace(0, float('nan')) * 100
)

# Keep categories with at least 100 views
eligible_categories = category_analysis[
    category_analysis['view'] >= 100
].copy()

# Select the top 10 by purchase-to-view event ratio
top_ratio_categories = eligible_categories.sort_values(
    by='purchase_view_ratio',
    ascending=False
).head(10)

print("Top 10 Categories by Purchase-to-View Event Ratio:")
display(
    top_ratio_categories[
        ['view', 'purchase', 'purchase_view_ratio']
    ].round(2)
)
```

```python
import matplotlib.pyplot as plt

ratio_data = top_ratio_categories.sort_values(
    by='purchase_view_ratio',
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    ratio_data.index,
    ratio_data['purchase_view_ratio']
)

plt.title('Top 10 Categories by Purchase-to-View Event Ratio')
plt.xlabel('Purchase-to-View Event Ratio (%)')
plt.ylabel('Product Category')

plt.tight_layout()
plt.savefig('category_conversion_ratios.png', dpi=300)
plt.show()
```

```python
# Summary of product prices by event type
price_analysis = df.groupby('event_type')['price'].agg(
    ['count', 'mean', 'median', 'min', 'max']
).round(2)

print("Product Price Analysis by Event Type:")
display(price_analysis)
```

```python
import matplotlib.pyplot as plt

price_means = df.groupby('event_type')['price'].mean()

# Keep events in a clear order
price_means = price_means.reindex(['view', 'cart', 'purchase'])

plt.figure(figsize=(8, 5))
plt.bar(price_means.index, price_means.values)

plt.title('Average Product Price by Event Type')
plt.xlabel('Event Type')
plt.ylabel('Average Price')

plt.tight_layout()
plt.savefig('average_price_by_event.png', dpi=300)
plt.show()
```

```python
# Extract the date from event_time
df['event_time'] = pd.to_datetime(df['event_time'])
df['date'] = df['event_time'].dt.date

# Count daily purchase events
daily_purchases = (
    df[df['event_type'] == 'purchase']
    .groupby('date')
    .size()
)

print("Daily Purchase Events:")
display(daily_purchases)

# Plot the daily trend
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(daily_purchases.index, daily_purchases.values, marker='o')

plt.title('Daily Purchase Events - October 2019 Sample')
plt.xlabel('Date')
plt.ylabel('Number of Purchase Events')
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('daily_purchase_trend.png', dpi=300)
plt.show()
```

```python
# Calculate key performance indicators

total_events = len(df)
total_sessions = df['user_session'].nunique()

view_events = (df['event_type'] == 'view').sum()
cart_events = (df['event_type'] == 'cart').sum()
purchase_events = (df['event_type'] == 'purchase').sum()

print("MARKETING FUNNEL - KEY PERFORMANCE INDICATORS")
print("-" * 45)
print("Total Events:", f"{total_events:,}")
print("Total Sessions:", f"{total_sessions:,}")
print("Product View Events:", f"{view_events:,}")
print("Add-to-Cart Events:", f"{cart_events:,}")
print("Purchase Events:", f"{purchase_events:,}")
print("View-to-Cart Event Rate:",
      round(cart_events / view_events * 100, 2), "%")
print("Purchase-to-View Event Rate:",
      round(purchase_events / view_events * 100, 2), "%")
print("Average Purchase Price:",
      round(df.loc[df['event_type'] == 'purchase', 'price'].mean(), 2))
```

```python
kpi_results = pd.DataFrame({
    'Metric': [
        'Total Events',
        'Total Sessions',
        'Product View Events',
        'Add-to-Cart Events',
        'Purchase Events',
        'View-to-Cart Event Rate (%)',
        'Purchase-to-View Event Rate (%)',
        'Average Purchase Price'
    ],
    'Value': [
        total_events,
        total_sessions,
        view_events,
        cart_events,
        purchase_events,
        round(cart_events / view_events * 100, 2),
        round(purchase_events / view_events * 100, 2),
        round(df.loc[df['event_type'] == 'purchase', 'price'].mean(), 2)
    ]
})

kpi_results.to_csv('marketing_funnel_kpis.csv', index=False)

print("KPI results saved successfully!")
display(kpi_results)
```
